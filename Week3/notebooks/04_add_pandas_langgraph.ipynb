````python
# LIBRARIES

from langchain_openai import ChatOpenAI
from langchain_community.utilities import SQLDatabase
from langchain.chains import create_sql_query_chain

from langgraph.graph import END, StateGraph

import os
import re
import yaml
from pprint import pprint
from typing import List, TypedDict

import pandas as pd
import sqlalchemy as sql

from IPython.display import Image
````

````python
# AI SETUP

os.environ["OPENAI_API_KEY"] = yaml.safe_load(open('../credentials.yml'))['openai']

OPENAI_LLM = ChatOpenAI(
    model = "gpt-4o-mini"
)

llm = OPENAI_LLM
````

````python
# SQL DATABASE SETUP

PATH_DB = "sqlite:///database/leads_scored.db"

sql_engine = sql.create_engine(PATH_DB)

conn = sql_engine.connect()
````

````python
def extract_sql_code(text):
    sql_code_match = re.search(r'```sql(.*?)```', text, re.DOTALL)
    sql_code_match_2 = re.search(r"SQLQuery:\s*(.*)", text)
    if sql_code_match:
        sql_code = sql_code_match.group(1).strip()
        return sql_code
    if sql_code_match_2:
        sql_code = sql_code_match_2.group(1).strip()
        return sql_code
    else:
        sql_code_match = re.search(r"sql(.*?)'", text, re.DOTALL)
        if sql_code_match:
            sql_code = sql_code_match.group(1).strip()
            return sql_code
        else:
            return None
````

````python
# * SQL Agent

db = SQLDatabase.from_uri(PATH_DB)

sql_generator = create_sql_query_chain(
    llm = llm,
    db = db,
    k = int(1e7)
)

sql_generator
````

````python
response = sql_generator.invoke({"question": "which 10 customers have the highest p1 probability of purchase?"})

pd.read_sql(extract_sql_code(response), conn)
````

````python
# * LANGGRAPH
class GraphState(TypedDict):
    """
    Represents the state of our graph.
    """
    question: str
    sql_query : str
    data: dict


def generate_sql(state):
    print("---GENERATE SQL---")
    question = state.get("question")
    
    # Generate SQL
    sql_query = sql_generator.invoke({"question": question})
    
    # Extract SQL code
    sql_query = extract_sql_code(sql_query)
    
    return {"sql_query": sql_query}

# New: Create Data Frame
def convert_dataframe(state):
    print("---CONVERT DATA FRAME---")

    sql_query = state.get("sql_query")
    
    df = pd.read_sql(sql_query, conn)
    
    return {"data": dict(df)}
  
    
def state_printer(state):
    """print the state"""
    print("---STATE PRINTER---")
    print(f"question: {state['question']}")
    pprint(f"SQL Query: {state['sql_query']}")
    print(f"Data: {pd.DataFrame(state['data'])}")
````

````python
# * WORKFLOW DAG

workflow = StateGraph(GraphState)

workflow.add_node("generate_sql", generate_sql)
workflow.add_node("convert_dataframe", convert_dataframe)
workflow.add_node("state_printer", state_printer)

workflow.set_entry_point("generate_sql")
workflow.add_edge("generate_sql", "convert_dataframe")
workflow.add_edge("convert_dataframe", "state_printer")
workflow.add_edge("state_printer", END)

app = workflow.compile()

Image(app.get_graph().draw_mermaid_png())
````

````python
# * TESTING

QUESTION = """
Which 10 customers have the highest p1 probability of purchase?
"""

inputs = {"question": QUESTION, "num_steps": 0}
for s in app.stream(inputs):
    print(s)
````
